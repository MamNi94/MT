```python
import serial
import time
from IPython.display import clear_output
```

```python
arduino = serial.Serial(port='COM5', baudrate=9600, timeout=.1)
#arduino = serial.Serial(port='/dev/ttyACM1', baudrate=9600, timeout=.1)

def set_servo(servo_id, angle):
    command = f"S,{servo_id},{angle}\n"
    arduino.write(command.encode())
```

while True:
    set_servo(0, 90)
    if arduino.in_waiting > 0:
        line = arduino.readline().decode().strip()
        #print("Arduino says:", line)

```python
def read_ir_data():
    if arduino.in_waiting > 0:
        line = arduino.readline().decode('utf-8').strip()
        if line.startswith("IR,"):
            try:
                
                parts = line.split(',')
                raw_value = int(parts[1])
       
                return raw_value
            except (IndexError, ValueError):
                return None
    return None
```

```python
while True:
    data = read_ir_data()
    if data:
       
        print(f"Raw: {data}")
        clear_output(wait=True) 
   
```

```python
#close gripper
def close_gripper():
    set_servo(0, 20)
    time.sleep(0.01)

    set_servo(3, 170)
    time.sleep(0.01)

    set_servo(4,60)
    
#open gripper
def open_gripper():
    set_servo(0, 110)
    time.sleep(0.01)

    set_servo(3, 80)
    time.sleep(0.01)

    set_servo(4, 150)
```

```python
#move angle right finger
#angle [80, 170] (opposite, corner)
set_servo(1, 80)
time.sleep(0.01)
#move angle left finger
#angle [160,70] (opposite, corner)
set_servo(2,160)
time.sleep(0.01)
#move angle corner
#angle [90] (middle) +-60
set_servo(5, 90)
```
