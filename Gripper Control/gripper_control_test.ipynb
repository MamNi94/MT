```python
import serial
import time
```

```python
arduino = serial.Serial(port='COM4', baudrate=9600, timeout=.1)
#arduino = serial.Serial(port='/dev/ttyACM1', baudrate=9600, timeout=.1)

def set_servo(servo_id, angle):
    command = f"S,{servo_id},{angle}\n"
    arduino.write(command.encode())
```

```python
#close gripper
def close_gripper():
    set_servo(0, 20)
    time.sleep(0.01)

    set_servo(3, 170)
    time.sleep(0.01)

    set_servo(4,60)
    
#open gripper
def open_gripper():
    set_servo(0, 110)
    time.sleep(0.01)

    set_servo(3, 80)
    time.sleep(0.01)

    set_servo(4, 150)
```

```python
#move angle right finger
#angle [80, 170] (opposite, corner)
set_servo(1, 80)
time.sleep(0.01)
#move angle left finger
#angle [160,70] (opposite, corner)
set_servo(2,160)
time.sleep(0.01)
#move angle corner
#angle [90] (middle) +-60
set_servo(5, 90)
```

```python
iteration = 0


while iteration <=5: 
    
    close_gripper()
   
    time.sleep(0.5)
    
    open_gripper()
    
    time.sleep(0.5)

    
    iteration +=1
  
```
